```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set(style="whitegrid")

# Make sure the output folder for plots exists
os.makedirs('../results/eda_visualizations', exist_ok=True)
```

```python
df = pd.read_csv('../data/raw/warsaw.csv')
df.head()
```

```python
df.info()
df['PRCP'].describe()
```

```python

prcp = df['PRCP'].dropna()

print(f"Skewness before transformation: {prcp.skew():.3f}")
print(f"Min: {prcp.min()}, Max: {prcp.max()}, Mean: {prcp.mean():.3f}, Median: {prcp.median():.3f}")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(prcp, bins=40, kde=True, color='steelblue')
plt.title('PRCP Distribution — Before Log Transformation')
plt.xlabel('Precipitation (PRCP)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/prcp_before_transformation.png', dpi=150)
plt.show()
```

```python
# log1p(x) = log(1 + x) -> safely handles PRCP = 0
df['PRCP_log'] = np.log1p(df['PRCP'])

df[['PRCP', 'PRCP_log']].head(10)
```

```python
prcp_log = df['PRCP_log'].dropna()

print(f"Skewness after transformation: {prcp_log.skew():.3f}")
print(f"Min: {prcp_log.min():.3f}, Max: {prcp_log.max():.3f}, Mean: {prcp_log.mean():.3f}, Median: {prcp_log.median():.3f}")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(prcp_log, bins=40, kde=True, color='seagreen')
plt.title('PRCP Distribution — After Log Transformation (log1p)')
plt.xlabel('log(1 + PRCP)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/prcp_after_transformation.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(prcp, bins=40, kde=True, color='steelblue', ax=axes[0])
axes[0].set_title('Before Log Transformation')
axes[0].set_xlabel('PRCP')

sns.histplot(prcp_log, bins=40, kde=True, color='seagreen', ax=axes[1])
axes[1].set_title('After Log Transformation')
axes[1].set_xlabel('log(1 + PRCP)')

plt.suptitle('PRCP Distribution: Before vs After Log Transformation')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/prcp_before_after_comparison.png', dpi=150)
plt.show()
```

```python
os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/warsaw_transformed.csv', index=False)
print('Saved transformed dataset to ../data/processed/warsaw_transformed.csv')
```
