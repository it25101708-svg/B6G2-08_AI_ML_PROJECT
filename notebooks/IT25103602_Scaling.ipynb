```python
import os
import pandas as pd

# Search recursively for warsaw.csv across parent directories
target_file = 'warsaw.csv'
found_path = None

for root, dirs, files in os.walk('.'):
  if target_file in files:
    found_path = os.path.join(root, target_file)
    break

if not found_path:
  for root, dirs, files in os.walk('..'):
    if target_file in files:
      found_path = os.path.join(root, target_file)
      break

if found_path:
  df = pd.read_csv(found_path)
  print(f"Successfully located and loaded dataset from: '{found_path}'")
  print(f'Dataset Dimensions: {df.shape}')
  display(df.head())
else:
  print('Current Directory Location:', os.path.abspath('.'))
  print('Files in current directory:', os.listdir('.'))
  raise FileNotFoundError(
      f"'{target_file}' is missing. Please download or copy 'warsaw.csv' into"
      ' this folder.'
  )
```
